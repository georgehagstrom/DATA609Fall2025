$$
A = U \Sigma V^T
$$

$A$ is $m\times n$,
$U$ is $m\times m$
$\Sigma$ $m \times n$
$V$ is $n \times n$

$U$ and $V$ are orthonormal matrices, which means:

$U U^T = I$, $V V^T = I$

$\Sigma$ is 0 except for the diagonals which are the singular values of $A$
